```python
#DAA PRACTICAL-1
import random
import time

def bubble_sort(a):
    n = len(a)

    for i in range(n - 1):
        for j in range(n - i - 1):
            if a[j] > a[j + 1]:
                a[j], a[j + 1] = a[j + 1], a[j]


def selection_sort(a):
    n = len(a)

    for i in range(n - 1):
        smallest = i
        for j in range(i + 1, n):
            if a[j] < a[smallest]:
                smallest = j

        a[i], a[smallest] = a[smallest], a[i]

def insertion_sort(a):
    n = len(a)
    for i in range(1, n):
        value = a[i]
        j = i - 1

        while j >= 0 and a[j] > value:
            a[j + 1] = a[j]
            j -= 1

        a[j + 1] = value

def merge(a, l, m, r):
    temp = []

    i = l
    j = m + 1

    while i <= m and j <= r:
        if a[i] < a[j]:
            temp.append(a[i])
            i += 1
        else:
            temp.append(a[j])
            j += 1

    while i <= m:
        temp.append(a[i])
        i += 1

    while j <= r:
        temp.append(a[j])
        j += 1

    for k in range(l, r + 1):
        a[k] = temp[k - l]


def merge_sort(a, l, r):
    if l >= r:
        return

    mid = l + (r - l) // 2

    merge_sort(a, l, mid)
    merge_sort(a, mid + 1, r)

    merge(a, l, mid, r)

def partition(a, low, high):
    pivot = a[high]
    pos = low

    for i in range(low, high):
        if a[i] < pivot:
            a[i], a[pos] = a[pos], a[i]
            pos += 1

    a[pos], a[high] = a[high], a[pos]

    return pos


def quick_sort(a, low, high):
    if low < high:
        p = partition(a, low, high)

        quick_sort(a, low, p - 1)
        quick_sort(a, p + 1, high)

def measure_simple(sort_func, arr):
    arr = arr.copy()
    start = time.perf_counter()
    sort_func(arr)
    stop = time.perf_counter()
    return int((stop - start) * 1_000_000)


def measure_recursive(arr, merge_flag):
    arr = arr.copy()
    start = time.perf_counter()
    if merge_flag:
        merge_sort(arr, 0, len(arr) - 1)
    else:
        quick_sort(arr, 0, len(arr) - 1)

    stop = time.perf_counter()
    return int((stop - start) * 1_000_000)

SIZE = 500

numbers = []
for i in range(SIZE):
    numbers.append(random.randint(0, 999))

names = [
    "Bubble Sort",
    "Selection Sort",
    "Insertion Sort",
    "Merge Sort",
    "Quick Sort"
]
times = [0] * 5
times[0] = measure_simple(bubble_sort, numbers)
times[1] = measure_simple(selection_sort, numbers)
times[2] = measure_simple(insertion_sort, numbers)
times[3] = measure_recursive(numbers, True)
times[4] = measure_recursive(numbers, False)
print("Number of Elements :", SIZE)
print()

for i in range(5):
    print(names[i], ":", times[i], "microseconds")

fastest = 0
slowest = 0
for i in range(1, 5):
    if times[i] < times[fastest]:
        fastest = i
    if times[i] > times[slowest]:
        slowest = i
print("\n---------------------------------")
print("Fastest Algorithm :", names[fastest])
print("Slowest Algorithm :", names[slowest])
print("---------------------------------")
```

```python
#DAA PRACTICAL-2
import time
def linear_search(arr, key):
    for i in range(len(arr)):
        if arr[i] == key:
            return i
    return -1

def binary_search(arr, key):
    low = 0
    high = len(arr) - 1
    while low <= high:
        mid = low + (high - low) // 2

        if arr[mid] == key:
            return mid
        elif arr[mid] < key:
            low = mid + 1
        else:
            high = mid - 1

    return -1

n = 100000
arr = list(range(1, n + 1))
key = int(input("Enter element to search: "))
start = time.perf_counter()
index = linear_search(arr, key)
stop = time.perf_counter()
print("\nLinear Search")
if index != -1:
    print("Element found at index", index)
else:
    print("Element not found")

print("Time Taken:", (stop - start) * 1_000_000, "microseconds")

start = time.perf_counter()
index = binary_search(arr, key)
stop = time.perf_counter()

print("\nBinary Search")

if index != -1:
    print("Element found at index", index)
else:
    print("Element not found")

print("Time Taken:", (stop - start) * 1_000_000, "microseconds")
```

```python
import time

def factorial_iterative(n):
    result = 1

    for i in range(1, n + 1):
        result *= i

    return result
def factorial_recursive(n):
    if n <= 1:
        return 1

    return n * factorial_recursive(n - 1)
n = int(input("Enter a non-negative integer : "))
if n < 0:
    print("Invalid input! Please enter a non-negative integer.")
else:
    start_iter = time.perf_counter()
    result_iter = factorial_iterative(n)
    end_iter = time.perf_counter()
    duration_iter = (end_iter - start_iter) * 1_000_000_000
    start_rec = time.perf_counter()
    result_rec = factorial_recursive(n)
    end_rec = time.perf_counter()
    duration_rec = (end_rec - start_rec) * 1_000_000_000
    print("\n--- Results for", n, "! ---")
    print("Iterative Result :", result_iter)
    print("Iterative Time   :", duration_iter, "ns")
    print("--------------------------------")
    print("Recursive Result :", result_rec)
    print("Recursive Time   :", duration_rec, "ns")
```
